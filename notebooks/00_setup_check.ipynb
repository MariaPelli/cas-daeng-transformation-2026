# Technischer Setup-Check

Dieses Notebook prüft nur, ob Python, die Kernbibliotheken und der Ausgangsdatensatz verfügbar sind. Es enthält noch keine fachliche Lösung.

```python
from pathlib import Path
import platform
import pandas as pd
import numpy as np
import pyarrow
import sklearn
import matplotlib
import seaborn as sns

def locate_root():
    for candidate in [Path.cwd(), *Path.cwd().parents]:
        if (candidate / 'data/input/trips_case_initial.parquet').exists():
            return candidate
    raise FileNotFoundError('Repository-Wurzel nicht gefunden')

ROOT = locate_root()
print('Python:', platform.python_version())
print('Repository:', ROOT)
```

```python
trips = pd.read_parquet(ROOT / 'data/input/trips_case_initial.parquet')
assert len(trips) == 200_900
assert len(trips.columns) == 24
assert trips['ingest_record_id'].notna().all()
print(f'Daten geladen: {len(trips):,} Zeilen × {len(trips.columns)} Spalten')
print('SETUP OK')
```
