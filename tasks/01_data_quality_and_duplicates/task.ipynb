## 0. Technische Vorbereitung

Die folgende Zelle sucht die Repository-Wurzel automatisch. Dadurch funktioniert dasselbe Notebook in JupyterLab, VS Code und GitHub Codespaces.

```python
from pathlib import Path
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 40)
sns.set_theme(style="whitegrid", context="notebook")

def find_repo_root(start: Path = Path.cwd()) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "data/input/trips_case_initial.parquet").exists():
            return candidate
    raise FileNotFoundError("Repository-Wurzel nicht gefunden.")

ROOT = find_repo_root()
DATA_PATH = ROOT / "data/input/trips_case_initial.parquet"

print(f"Repository: {ROOT}")
print(f"Datensatz:  {DATA_PATH}")
```

```python
trips = pd.read_parquet(DATA_PATH)

assert DATA_PATH.exists()
assert "ingest_record_id" in trips.columns
assert "tpep_pickup_datetime" in trips.columns
assert trips["ingest_record_id"].notna().all()

print("Datensatz erfolgreich geladen.")
print(f"Dimensionen: {trips.shape}")
```
